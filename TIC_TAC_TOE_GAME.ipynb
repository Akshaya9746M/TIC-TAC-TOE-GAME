```python

board = [" " for i in range(9)]

def display_board():
    print()
    print(board[0] + " | " + board[1] + " | " + board[2])
    print("--+---+--")
    print(board[3] + " | " + board[4] + " | " + board[5])
    print("--+---+--")
    print(board[6] + " | " + board[7] + " | " + board[8])
    print()

def check_winner():
    winning_positions = [
        [0, 1, 2],
        [3, 4, 5],
        [6, 7, 8],
        [0, 3, 6],
        [1, 4, 7],
        [2, 5, 8],
        [0, 4, 8],
        [2, 4, 6]
    ]

    for pos in winning_positions:
        if (board[pos[0]] == board[pos[1]] ==
            board[pos[2]] != " "):
            return True

    return False

def play_game():
    current_player = "X"

    while True:
        display_board()

        try:
            choice = int(input(
                f"Player {current_player}, choose position (1-9): "
            ))
        except ValueError:
            print("Please enter a valid number!")
            continue

        if choice < 1 or choice > 9:
            print("Choose a number between 1 and 9!")
            continue

        if board[choice - 1] != " ":
            print("This position is already taken!")
            continue

        board[choice - 1] = current_player

        if check_winner():
            display_board()
            print("Congratulations! Player", current_player, "wins!")
            break

        if " " not in board:
            display_board()
            print("Game Draw!")
            break

        current_player = "O" if current_player == "X" else "X"

play_game()
```
